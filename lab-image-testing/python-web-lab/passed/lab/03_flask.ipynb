# 03 เขียน route ใน Flask

เว็บของแลปนี้รันจาก `app.py` ที่ port **5000** และ reload เองทุกครั้งที่ save ไฟล์

1. เปิด `app.py` ในแถบซ้าย
2. เพิ่ม route `/square/<int:n>` ให้ตอบ JSON `{"n": n, "square": n * n}`
3. save แล้วรันเซลล์ข้างล่าง

```python
# ตรวจคำตอบ: เรียกเว็บของแลปจากใน container
import json
import urllib.request

with urllib.request.urlopen("http://127.0.0.1:5000/square/12", timeout=5) as response:
    body = json.load(response)

assert body == {"n": 12, "square": 144}, body
print("ผ่าน")
```
